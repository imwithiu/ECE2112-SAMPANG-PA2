# A. REPRODUCABLE NORMALIZATION PROBLEM


```python
import numpy as np
```

```python
np.random.seed(2112)
```

```python
X = np.random.randint(10, 101, size=(5,5))
```

```python
mean = X.mean()
std = np.std(X)
```

```python
X_normalized = (X - X.mean())/std
```

```python
mean_normalized = (X_normalized.mean())
std_normalized = np.std(X_normalized)
```

*PROBLEM CHECK A:*

```python
print("X:\n",X)
print("\nX_normalized:\n",X_normalized)
print("\nmean:\n",mean)
print("\nStandard Deviation:\n",std)
print("\nNormalized mean:\n",mean_normalized)
print("\nNormalized Standard Deviation:\n",std_normalized)
```

```python
np.save("X_normalized.npy", X_normalized)
```

# B. CUBES DIVISIBLE BY 4

```python
integers = np.arange(1,101)
```

```python
cubed_integers = integers**3
```

```python
C = cubed_integers.reshape(10,10)
```

```python
div_by_4 = C[C%4 == 0]
```

*PROBLEM CHECK B:*

```python
print("The shape of C: \n",C.shape)
print("\nThe array divisible by 4: \n", div_by_4)
print("\nNumber of elements in div_by_4: \n",div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

#C. ABOVE-MEAN SQUARES PROBLEM

```python
bintegers = np.arange(1,37)
```

```python
S = bintegers**2
```

```python
S = np.reshape(S,(6,6))
```

```python
S_mean = S.mean()
```

```python
above_mean = S[S>S_mean]
```

*PROBLEM CHECK C:*

```python
print("S: \n",S)
print("\nS_mean:\n",S_mean)
print("\nabove_mean:\n",above_mean)
print("\nThe number of selected elements: \n",above_mean.size)
```

```python
np.save("above_mean.npy", above_mean)
```
